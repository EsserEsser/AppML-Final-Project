```python
path = "..\\Data"

#For each file in the path extract number after last underscore and before .jpg
import os
pops = []
for file in os.listdir(path):
    if file.endswith(".jpg"):
        number = file.split("_")[-1].split(".")[0]
        pops.append(int(number))
print(pops)
        

```
