```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
#Make a horizontal bar chart using the following data:
#Model 1 (Quick): R^2 = 0.663
#Model 2 (Trans): R^2 = 0.477
#Model 3 (Reg+Class): R^2 = 0.743
#Model 4 R^2 = 0.420
#Model 5 R^2 = 0.420

colors = [
    "#1D3557",  # Deep navy blue (great for primary data)
    "#F4A261",  # Warm sandy mustard
    "#2A9D8F",  # Deep teal/green
#    "#E63946",  # Punchy brick red
    "#457B9D",  # Muted steel blue
    "#E76F51",  # Terracotta
    "#A8DADC",  # Soft pale blue (for secondary/background data)
]


models = ['Quick CNN', 'Trans (300m)', 'Reg+Class (100m)', 'Model 4', 'Model 5']
r2_scores = [0.663, 0.477, 0.743, 0.420, 0.420]
plt.figure(figsize=(8, 4))
plt.barh(models, r2_scores, color=colors,height=0.5,edgecolor='black' )
plt.xlabel(r'$R^2$ Score')
plt.title(r'$R^2$ Scores of Different Models')
plt.gca().invert_yaxis()  # Highest R^2 at the top
plt.tight_layout()
#plt.grid(axis='x')
plt.show()
```
