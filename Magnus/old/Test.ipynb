```python
import zipfile
import os

files = [
    'jpegs_all_over_6755.zip',
    'jpegs_all_under_equal_6755.zip'
]

for file in files:
    print(f"\nProcessing {file}...")
    with zipfile.ZipFile(f'/content/drive/MyDrive/AppML_Data/{file}', 'r') as zip_ref:
        jpegs_only = [f for f in zip_ref.namelist() if os.path.splitext(f)[1] == '.jpg']
        total = len(jpegs_only)
        print(f"Found {total} jpg files to extract")

        for i, jpg in enumerate(jpegs_only, 1):
            zip_ref.extract(jpg, '/content/drive/MyDrive/AppML_Data/Data_All')
            if i % 1000 == 0:
                print(f"  {i}/{total} extracted ({(i/total)*100:.1f}%)")

    print(f"✅ Done with {file}")

print("\n🎉 All files extracted!")
```
